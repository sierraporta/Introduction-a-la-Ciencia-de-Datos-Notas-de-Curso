# 1. Python como herramienta

Antes de trabajar con tablas, recordemos algunas piezas mínimas de Python.


```python
nombre = "Ciencia de Datos"
semestre = 1
temperatura = 28.5

print(nombre)
print(semestre)
print(temperatura)
```

```python
temperaturas = [27, 29, 28, 30, 31, 29]

print("Número de observaciones:", len(temperaturas))
print("Mínimo:", min(temperaturas))
print("Máximo:", max(temperaturas))
print("Promedio:", sum(temperaturas) / len(temperaturas))
```

### Pregunta rápida

¿Qué ocurre cuando una lista de Python empieza a contener miles o millones de valores y necesitamos hacer cálculos repetidamente?

Aquí aparece **NumPy**.


# 2. NumPy: pensar en arreglos


```python
temperaturas_np = np.array(temperaturas)

temperaturas_np
```

```python
print("Promedio:", temperaturas_np.mean())
print("Desviación estándar:", temperaturas_np.std())
print("Mínimo:", temperaturas_np.min())
print("Máximo:", temperaturas_np.max())
```

```python
temperaturas_np - temperaturas_np.mean()
```

La última operación es importante: NumPy permite aplicar una operación a todo el arreglo de manera compacta y eficiente.

No necesitamos recorrer manualmente cada elemento para restarle el promedio.

# 6. Estadística descriptiva


```python
mean_pm25 = df["pm25"].mean()
median_pm25 = df["pm25"].median()
std_pm25 = df["pm25"].std()

print("Promedio:", mean_pm25)
print("Mediana:", median_pm25)
print("Desviación estándar:", std_pm25)
```

### ¿Por qué mirar más de una medida?

El promedio resume, pero no cuenta toda la historia.

Podemos comparar:

- promedio;
- mediana;
- mínimo y máximo;
- dispersión.

Más adelante estudiaremos estas ideas formalmente.

# 9. Visualización: convertir datos en una imagen


```python
plt.figure(figsize=(10, 5))

plt.bar(df["city"], df["pm25"])

plt.xlabel("Ciudad")
plt.ylabel("PM2.5")
plt.title("PM2.5 en ciudades colombianas — datos demostrativos")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Pregunta para discutir

**¿Qué nos permite ver la gráfica que era más difícil de ver en la tabla?**

La visualización no sustituye al análisis. Nos ayuda a encontrar patrones, diferencias y posibles preguntas.


# 10. Buscar relaciones


```python
plt.figure(figsize=(7, 5))

plt.scatter(df["temperature_c"], df["pm25"])

plt.xlabel("Temperatura (°C)")
plt.ylabel("PM2.5")
plt.title("Temperatura vs. PM2.5 — datos demostrativos")

plt.tight_layout()
plt.show()
```

### Importante

Una gráfica puede sugerir una relación, pero **ver una tendencia no demuestra causalidad**.

En próximas clases aprenderemos herramientas para estudiar relaciones de manera más rigurosa.


# 11. Una pequeña transformación con NumPy


```python
df["pm25_centered"] = df["pm25"] - df["pm25"].mean()

df[["city", "pm25", "pm25_centered"]]
```

Hemos creado una nueva variable a partir de una variable existente.

Esto es una primera aproximación a la idea de **transformación y construcción de variables**.
